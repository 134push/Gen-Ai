```python
## magic method are alsso called as dunder methods 
# because they have double underscore before and after their name
# and they are specials method that allows to customize how objects behave 

# thses method help us out define: 
#  1. object inetializations (__init__)
#  2. string representation (__str__,__repe__)
#  3. arithimatic operations (__add__,__sub__,etc)
#  4. compariosn operations (eq, lt) 
# 5. length,indexing,iterations {len,getitem,etc)
```

```python
# why to use magic methods 

## 1. makess objects behave likes built in types (eq . lists , integers)
## 2. improve redablities by allowing basic opertaions abjects1 + object2 
## 3. simplifies code by replacing verbose functions call with natural syntax .

# print(object) --> str --> what is return when 
# print(object) --> is used 

# arithmatics operations can be done among the objects by using magic methods 
```

```python
class student:
    def __init__(self,roll_no,marks):
        self.roll_no = int(roll_no)
        self.marks = int(marks)

    def __add__(self,mark):
        return student(self.roll_no  + mark.roll_no,self.marks + mark.marks)

    def  __str__(self):
        return f"{self.roll_no} with marks {self.marks}"

p1 = student(123,23)
p2 = student(124,25)

all1 = p1 + p2 
print(p1)
```

```python
# operator overloading 

## operator overloading allows us to define how operators [+,-,/,%,*,etc]
#  makes custom objects behave like built in types  
# to improve code readablites by using natrul synatax (object1 + object2 )

# operator --> + --> method name --> __add__(self,other)
## just like this __sub__,__mul__,__truediv__(# division),__eq__(self,other)(# for ==),
# __gt__(self,other)(# > )

```

```python
class Bank:
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance

    def __add__(self,add):
        return Bank(self.name + " & " + add.name,self.balance + add.balance)

    def __sub__(self,sub):
        return Bank(self.name,self.balance - sub.balance)    

    def __mul__(self,mul):
        return Bank(self.name,self.balance * mul.balance)

    def __str__(self):
        return f"{self.name} with balance {self.balance}"

b1 = Bank("pushpam",20000)
b2 = Bank("raj",15000)
      
team = b1 + b2 
print(team)      
```
