```python
#self -> it refers the current object of the class
#constructor -> a constructor is being referred by init it called automatics called 
# even if user forget to create a constructor still constructor will going with their methods

# 1 parameterized constructor -> it consiste of parameters inside it that every object has specified attributes associated to it 
class cartoon:
    def __init__(self,name,show_name,friends):
        self.name = name
        self.show_name = show_name
        self.friends = friends

    def introduction(self):
        print(f"i am {self.name} .")
        print(f"i am working at {self.show_name} .")
        print(f"my friend is {self.friends} .") 

nobita = cartoon(input("enter the name : "),"doremon","jiyan")   

nobita.introduction()
```

```python
# types of methods

#1.instance methods(object method) -> instance method operate on object attributes .
## the instance method going to use self keyword to access and modify object attributes .
## this method are called using an object .

class laptop:
    def __init__(self,name,models,cpu_name):
        self.name = name
        self.models = models
        self.cpu_name = cpu_name

    def details(self):
        print(f"the laptop name is {self.name} and model is {self.models} and it have {self.cpu_name} procsser .")

asus = laptop("a15","asus","rtx3050")

asus.details()
```

```python
#2.class methods ->  it is the method of the whole class or the whole blueprint 
# which will remain the same for all of the objects then such kind of method is being reffered
# in class method , these methods work with class level attributes .
# they are called using the class name as well 
# for specific that current method in a class   

class bike:

    bike_type = "2-wheeler"
    def __init__(self,name):
        self.name = name

    @classmethod
    def main(cls):
        print(f"my bike type is {cls.bike_type}")

    def info(self):
        print(f"the bike name is {self.name}")

bullet = bike("hunter") 
bullet.main()
bullet.info()
```
