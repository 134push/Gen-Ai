```python
# types of polymorphism

# 1. method overiding --> a child class provides a specifics implementation of a method taht has already been defined into the parent class .
# this allow dynamic selction at a runtime .

class car:
    def model(self):
        print("this is bmw model")
class bike(car):
    def model(self):
        print("this is honda model")

class vehical(car):
    def model(self):
        print("this is vehical model")        

def new_model(objects):
    print(objects.model())

car = car()
new_model(car)    

bike = bike()
new_model(bike)
```

```python
class shape:
    def area(self):
        return "this method returns the area of the shape"
class rectangle(shape):
    def __init__(self,width,hight):
        self.width = width
        self.hight = hight
    
    def area(self):
        return self.width*self.hight

class circle(shape):
    def __init__(self,redius):
        self.reduis = reduis

    def area(self):
        return 3.14*self.reduis*self.reduis

shape = shape()
shape.area()         

rectangle = rectangle(6,4)
rectangle.area()
```
