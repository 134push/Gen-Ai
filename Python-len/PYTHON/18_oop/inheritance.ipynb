```python
## inheritance is a fundamental concept in oop because of which the child class is able to inherit the properties or attributes and functionalities of the parent class 

## types of inheritance 
# 1. single inheritance 
## the features or the attributes from one parents class will be gatting inherited into one child class 

class student:
    def __init__(self,name,roll_no):
        self.name = name 
        self.roll_no = int(roll_no)
    def info(self):
        print(f"hello my name is {self.name} and my roll no is {self.roll_no}")

class subject(student):
    def __init__(self,name,roll_no,subject):
        super().__init__(name,roll_no)
        self.subject = subject
    def info(self):
        print(f"my name is {self.name} and my roll no is {self.roll_no} . i am currently studying {self.subject}")

student1 = subject("pushpam",101,"maths")

student1.info()
```

```python
# 2. multilevel inheritance
## grandparents -> parents -> child 

class collage:
    def __init__(self,collage_name,pincode):
        self.collage_name = collage_name
        self.pincode = int(pincode)
class student(collage):
    def __init__(self,collage_name,pincode,name,roll_no):
        super().__init__(collage_name,pincode)
        self.name = name 
        self.roll_no = int(roll_no)
    def inf(self):
        print(f"hello my name is {self.name} and my roll no is {self.roll_no}")

class subject(student):
    def __init__(self,collage_name,pincode,name,roll_no,subject):
        super().__init__(collage_name,pincode,name,roll_no)
        self.subject = subject
    def info(self):
        print(f"i am currently studying at {self.collage_name}-{self.pincode} . ")

student1 = subject("mbit",388120,"pushpam",101,"maths")

student1.info()
student1.inf()
```

```python
# 3. multiple inheritance
# in multiple inheritance the properties of two parent class or multipls parent class will be getting inherited into  the child class 

class mobile:
    def __init__(self,model_name,id):
        self.model_name = model_name
        self.id = int(id)

    def mobile_info(self):
        print(f"mobile model name is {self.model_name} and id is {self.id}")    

class laptop:
    def __init__(self,laptop_model_name,id_no):
        self.laptop_model_name = laptop_model_name
        self.id_no = int(id_no)

    def mobile_info(self):
        print(f"laptop model name is {self.laptop_model_name} and id_no is {self.id_no}")

class elctronic(mobile,laptop):
    def __init__(self,model_name,id,laptop_model_name,id_no,date):
        mobile.__init__(self, model_name, id)
        laptop.__init__(self, laptop_model_name, id_no)
        self.date = int(date)

    def details(self):
        print(f"the mobile model name is {self.model_name}-{self.id}")
        print(f"the laptop model name is {self.laptop_model_name}-{self.id_no}")

asus = elctronic("asus",123,"asus f16",562,2026)

asus.details()


   
```
