```python
## encapsulation is a way of wrapping data (varibles or methods ) inside a capsule or a class as a single uniit .
## IT RESTRICTS THE DIRECT ACCESS TO SOME OF THE OBJECTS COMPONENTS WHICH IS A MEANS OF PREVENTING ACCIDENTAL INFERENCE AND MISUES OF THE DATA 
```

```python
## public members --> accessible from anwhere or attributes are accessible outside of the classs 
## this members are accssible by the derived class
## public members attributes is directaly declared 

## dir() --> it takes the input as the object name and give us this idea that which member is public , private or protected 

class Car:
    def __init__(self,model,name):
        self.model = model 
        self.name =  name 

    def details(self):
        print(f"the model is {self.model} and car name is {self.name}")

car = Car("bmw","m5")

car.details()

dir(car)
```

```python
## private members --> a private member is not accessile outside the class and not even in the derived class and in order to have the access over these private member or 
## to bring some modifications on these members we use getter and setter method 

class Car:
    def __init__(self,model,name,id):
        self.model = model
        self.name = name 
        self.__id = id 

car = Car("bmw","m5",2356)

car.name
dir (Car)

```

```python
## getter and setter function --> it will be helping us out with accessing the private data 

class Car1:
    def __init__(self,model,name,id):
        self.model = model
        self.name = name 
        self.__id = id 

    ## getter method 
    def accessing_add(self):
        print(f"the model is {self.model} with id is {self.__id}")
 
    def details(self,new_id):
        self.__id = new_id
        print(f"your new id is {self.__id}")
car1 = Car1("bmw","m5",2356)

car1.details(859)
car1.accessing_add()



    
```

```python
## protected data is the data which is not accessible outside the class but the protected data is going to be accessible by the derived class or child class


class Car1:
    def __init__(self,model,name,id,company_id):
        self.model = model
        self.name = name 
        self._company_id = company_id 
        self.__id = id 

    ## getter method 
    def accessing_add(self):
        print(f"the model is {self.model} with id is {self.__id} . the company_id is {self._company_id}")
 
    def details(self,new_id):
        self.__id = new_id
        print(f"your new id is {self.__id}")

class Company(Car1):
    def __init__(self,model,name,id,company_id):
        super().__init__(model,name,id,company_id)
    def show(self):
        print(f"the comany id is {self._company_id}")

car1 = Car1("bmw","m5",2356,789)
company1 = Company("bmw","m5",2356,789)
car1.details(859)
car1.accessing_add()

company1.show()


    
```
