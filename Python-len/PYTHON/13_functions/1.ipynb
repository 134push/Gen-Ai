```python
import time

def counting():
    for i in range(1, 101):  # 1 થી 100 સુધી કાઉન્ટિંગ
        print(f"\rCounting: {i}", end="", flush=True)
        time.sleep(0.05)     # કાઉન્ટિંગ સ્પીડ (0.05 સેકન્ડ)
    
    print()                  # લૂપ પૂરી થયા પછી નવી લાઇન માટે
    return i                 # છેલ્લો નંબર રિટર્ન કરશે

# Function Call
final_no = counting()
print(f"Your counting no is = {final_no}")
```
