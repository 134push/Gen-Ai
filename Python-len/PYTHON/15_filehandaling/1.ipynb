```python
def count(file_path):
    with open(file_path, "r") as f:
        lines = f.readlines()
        total_lines = len(lines)
        total_words = sum(len(line.split()) for line in lines)
        total_characters = sum(len(line) for line in lines)
        
    return total_characters, total_lines, total_words

# Example usage:
f = "1.txt"
characters, lines, words = count(f)

print(f"Characters: {characters}")
print(f"Lines: {lines}")
print(f"Words: {words}")
```
