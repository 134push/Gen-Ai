# Python Dictionary - 5 Students Nested Dictionary
This notebook demonstrates how to create and operate on a dictionary containing 5 student dictionaries (nested dictionary).

```python
# 1. Create a dictionary containing 5 student dictionaries
students = {
    "student1": {
        "id": 101,
        "name": "Rahul Sharma",
        "age": 20,
        "branch": "Computer Science",
        "cgpa": 8.8
    },
    "student2": {
        "id": 102,
        "name": "Priya Patel",
        "age": 21,
        "branch": "Information Technology",
        "cgpa": 9.2
    },
    "student3": {
        "id": 103,
        "name": "Amit Kumar",
        "age": 20,
        "branch": "Electronics",
        "cgpa": 7.9
    },
    "student4": {
        "id": 104,
        "name": "Neha Singh",
        "age": 22,
        "branch": "Mechanical",
        "cgpa": 8.5
    },
    "student5": {
        "id": 105,
        "name": "Pushpam Anand",
        "age": 21,
        "branch": "Artificial Intelligence",
        "cgpa": 9.6
    }
}

print("Nested Students Dictionary Created Successfully:")
print(students)
```

```python
# 2. Access data of individual students
print("Student 1 Name:", students["student1"]["name"])
print("Student 5 Branch:", students["student5"]["branch"])
print("Student 2 CGPA:", students["student2"]["cgpa"])
```

```python
# 3. Iterate over all 5 students using items()
print("=" * 40)
print("     ALL STUDENT RECORDS")
print("=" * 40)

for std_key, std_info in students.items():
    print(f"\n{std_key.upper()}:")
    for key, value in std_info.items():
        print(f"  {key.title()}: {value}")
    print("-" * 30)
```

```python
# 4. Update data for a specific student
students["student3"]["cgpa"] = 8.2
print("Updated Student 3 CGPA:", students["student3"]["cgpa"])
```

```python
# 5. Add a new field (email) to a student dictionary
students["student5"]["email"] = "pushpam@gmail.com"
print("Student 5 details after adding email:")
print(students["student5"])
```

```python
# 6. Total count of students in the dictionary
print("Total Students:", len(students))
```
